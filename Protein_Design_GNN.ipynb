```python
import urllib.request
```

```python
pdb_id = "1N5U"
```

```python
url = f"https://files.rcsb.org/download/{pdb_id}.pdb"

print(url)
```

```python
pdb_file = f"data/pdb/{pdb_id}.pdb"

print(pdb_file)
```

```python
urllib.request.urlretrieve(url, pdb_file)

print("Protein downloaded successfully!")
print("Saved at:", pdb_file)
```

```python
with open(pdb_file, "r") as file:
    for i in range(20):
        line = file.readline()
        print(line.strip())
```

```python
atom_count = 0

with open(pdb_file, "r") as file:
    for line in file:

        if line.startswith("ATOM"):
            print(line.strip())

            atom_count += 1

            if atom_count == 10:
                break
```
